```python
import pandas as pd  

# Load Apple stock CSV
df = pd.read_csv('downloads/HistoricalQuotes.csv')  

# Clean up column names (remove spaces)
df.columns = df.columns.str.strip()

# Rename 'Close/Last' → 'Close'
df.rename(columns={"Close/Last": "Close"}, inplace=True)

# Fill missing values with 0
df.fillna(0, inplace=True)

# Sort by closing price (highest first)
df.sort_values("Close", ascending=False, inplace=True)

# Show first 5 rows
df.head()
```
