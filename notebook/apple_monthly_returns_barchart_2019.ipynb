```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("downloads/HistoricalQuotes.csv")

# --- Data Cleaning ---
df.columns = df.columns.str.strip()
df.rename(columns={"Close/Last": "Close"}, inplace=True)

# Clean 'Close' column (remove $ and commas → float)
df["Close"] = (
    df["Close"].astype(str)
    .str.replace(r"[\$,]", "", regex=True)
    .astype(float)
)

# Convert Date column to datetime, sort, set index
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df = df.dropna(subset=["Date"]).sort_values("Date").set_index("Date")

# --- Calculate Returns ---
df["Return"] = df["Close"].pct_change()

# --- Resample to Monthly (Mean Returns) ---
monthly_returns = df["Return"].resample("M").mean()

# Focus on 2019 only
monthly_2019 = monthly_returns["2019"]

# --- Plot ---
plt.figure(figsize=(10, 5))
monthly_2019.plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Apple Average Monthly Returns (2019)", fontsize=14, weight="bold")
plt.ylabel("Average Return", fontsize=12)
plt.xlabel("Month", fontsize=12)

# Format x-axis as month names
plt.xticks(range(len(monthly_2019.index)), 
           [d.strftime("%b") for d in monthly_2019.index], 
           rotation=0, fontsize=10)

plt.axhline(0, color="black", linewidth=0.8)  # baseline
plt.tight_layout()
plt.show()
```
