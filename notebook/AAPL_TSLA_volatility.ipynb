```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf

tickers = ["AAPL", "TSLA"]
start = "2020-01-01"

#Download files
prices = yf.download(tickers, start=start, auto_adjust=True, progress=False)["Close"].dropna()

#Daily returns
rets = prices.pct_change().dropna()

#Cumulative returns chart
cum = (1 + rets).cumprod()
plt.figure(figsize=(10,5))
cum.plot(ax=plt.gca())
plt.title("Cumulative Returns — AAPL vs TSLA")
plt.xlabel("Year"); plt.ylabel("Growth of $1"); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()

# volatility compare (annualized)
overall_vol = (rets.std() * np.sqrt(252)).round(3)
print("\nAnnualized Volatility (overall):")
print(overall_vol)
```
