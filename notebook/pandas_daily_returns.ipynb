```python
import pandas as pd

df = pd.read_csv("downloads/HistoricalQuotes.csv")

print(df.columns)
```

```python
# Remove spaces from columns
df.columns = df.columns.str.strip()

# Rename Close/Last to Close
df.rename(columns={"Close/Last": "Close"}, inplace=True)

# Cleaning the close column before calculation (removed (,) and changed text to numbers)
df["Close"] = (
    df["Close"].astype(str)
                .str.replace(r"[\$,]", "", regex=True)
                .str.strip()
                .astype(float)
)

# Sort by date and set date as index
df = df.sort_values("Date")

#Apply the equation
df["Return"]=df["Close"].pct_change()

# check
print(df[["Close", "Return"]].head())
```
