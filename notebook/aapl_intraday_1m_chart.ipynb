```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf

ticker = "AAPL"
df = yf.download(ticker, period="1d", interval="1m",
                 auto_adjust=True, progress=False)

#Close series
price = df[["Close"]].dropna().astype(float).squeeze("columns")
price.name = ticker

# Latest bar
last_time  = price.index[-1]
last_price = price.iloc[-1]   # scalar
print(f"Last point: {last_time} ${last_price:.2f}")

# Plot
ax = price.plot(figsize=(10,4), title=f"{ticker} — Intraday (1m, Yahoo ~15m delayed)")
ax.set_xlabel("Time"); ax.set_ylabel("Price ($)")
plt.tight_layout(); plt.show()
```
